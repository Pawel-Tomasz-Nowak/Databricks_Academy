```python
dbutils.widgets.text("bronze_catalog","dbr_dev")
dbutils.widgets.text("bronze_schema","artemzharkov10_bronze")

BRONZE_CATALOG = dbutils.widgets.get("bronze_catalog")
BRONZE_SCHEMA = dbutils.widgets.get("bronze_schema")
```

```python
spark.sql(f"CREATE DATABASE IF NOT EXISTS {BRONZE_CATALOG}.{BRONZE_SCHEMA}")
# print(f"Created schema: {BRONZE_SCHEMA}")

spark.sql(f"CREATE VOLUME IF NOT EXISTS {BRONZE_CATALOG}.{BRONZE_SCHEMA}.raw_data")
# print(f"Raw_data was created")
```

```python
LOG_TABLE_NAME = f"{BRONZE_CATALOG}.{BRONZE_SCHEMA}.gdelt_download_logs"

spark.sql(f"""
    CREATE TABLE IF NOT EXISTS {LOG_TABLE_NAME} (
        filename STRING,
        downloaded_at TIMESTAMP
    )""")
```

```python
spark.sql(f"""
    CREATE TABLE IF NOT EXISTS {BRONZE_CATALOG}.{BRONZE_SCHEMA}.stream_bitcoin_bronze (
        Date STRING,
        Open STRING,
        High STRING,
        Low STRING,
        Close STRING,
        Volume STRING,
        _rescued_data STRING,
        source_filename STRING,
        ingest_timestamp TIMESTAMP
    )
""")
```
