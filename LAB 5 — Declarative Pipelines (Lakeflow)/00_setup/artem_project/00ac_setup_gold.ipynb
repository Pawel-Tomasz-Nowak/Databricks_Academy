```python
dbutils.widgets.text("gold_catalog","dbr_dev")
dbutils.widgets.text("gold_schema","artemzharkov10_gold")

GOLD_CATALOG = dbutils.widgets.get("gold_catalog")
GOLD_SCHEMA = dbutils.widgets.get("gold_schema")
```

```python
spark.sql(f"CREATE DATABASE IF NOT EXISTS {GOLD_CATALOG}.{GOLD_SCHEMA}")
# print(f"Created schema: {GOLD_SCHEMA}")
```

```python
spark.sql(f"""
CREATE TABLE IF NOT EXISTS {GOLD_CATALOG}.{GOLD_SCHEMA}.bitcoin_news_gold (
    NewsDateClean DATE,
    URL STRING,
    Actor1Name STRING,
    Actor2Name STRING,
    NewsTone DOUBLE,
    AverageTone DOUBLE,
    ActionGeo_Lat DOUBLE,
    ActionGeo_Long DOUBLE
)
""")

```

```python
spark.sql(f"""
CREATE TABLE IF NOT EXISTS {GOLD_CATALOG}.{GOLD_SCHEMA}.finance_bitcoin_anomaly_gold (
    AnomalyDate DATE,
    PriceOnAnomalyDay DOUBLE,
    BlockAveragePrice DOUBLE,
    DeviationPercentage DOUBLE
)
""")
```

```python
spark.sql(f"""
CREATE TABLE IF NOT EXISTS {GOLD_CATALOG}.{GOLD_SCHEMA}.bitcoin_same_day_mart (
    AnomalyDate DATE,
    PriceOnAnomalyDay DOUBLE,
    DeviationPercentage DOUBLE,
    NewsCountOnAnomalyDay BIGINT,
    AverageToneOnDay DOUBLE
)
""")
```

```python
spark.sql(f"""
CREATE TABLE IF NOT EXISTS {GOLD_CATALOG}.{GOLD_SCHEMA}.stream_bitcoin_anomaly_gold (
    AnomalyDate TIMESTAMP,
    PriceOnAnomalyDay DOUBLE,
    MarketCapOnAnomalyDay DOUBLE,
    BlockAveragePrice DOUBLE,
    DeviationPercentage DOUBLE
)
""")
```
