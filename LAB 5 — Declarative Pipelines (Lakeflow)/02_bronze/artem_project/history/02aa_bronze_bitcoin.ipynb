```python
from pyspark.sql.types import StructType, StructField, StringType, DoubleType, DateType
from pyspark.sql.functions import input_file_name, current_timestamp, current_date,col
```

```python
dbutils.widgets.text("catalog","dbr_dev")
dbutils.widgets.text("schema","artemzharkov10_bronze")

CATALOG = dbutils.widgets.get("catalog")
SCHEMA = dbutils.widgets.get("schema")
```

```python
SOURCE_VOLUME_PATH =f"/Volumes/{CATALOG}/{SCHEMA}/raw_data/finance_history/BITCOIN.csv"

market_schema = StructType([
    StructField("Date", StringType(), True), 
    StructField("Close", DoubleType(), True),
    StructField("High", DoubleType(), True),
    StructField("Low", DoubleType(), True),
    StructField("Open", DoubleType(), True),
    StructField("Volume", DoubleType(), True)
])
```

```python

df_bronze = (spark.read.format("csv")
  .option("header", "true")
  .schema(market_schema)
  .load(SOURCE_VOLUME_PATH)) 
df_bronze_with_metadata = (
  df_bronze
    .withColumn("source_filename", col("_metadata.file_path"))
    .withColumn("ingest_timestamp", current_timestamp())
    .withColumn("load_date", current_date())
)


(df_bronze_with_metadata.write
  .format("delta")
  .mode("overwrite")
  .option("overwriteSchema", "true")
  .saveAsTable(f"{CATALOG}.{SCHEMA}.finance_bitcoin_bronze"))

# display(df_bronze_with_metadata.limit(30))
```
