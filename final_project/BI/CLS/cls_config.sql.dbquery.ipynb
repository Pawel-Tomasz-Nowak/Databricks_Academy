```sql
CREATE OR REPLACE FUNCTION IDENTIFIER(:catalog_name).IDENTIFIER(:schema_name).mask_video_id(raw_id STRING)
RETURN 
    CASE 
        WHEN is_account_group_member('admins') THEN raw_id
        ELSE CONCAT('VID-***-', RIGHT(raw_id, 3))
    END;


CREATE OR REPLACE VIEW IDENTIFIER(:catalog_name).IDENTIFIER(:schema_name).cls_dim_music_metadata AS
SELECT 
    IDENTIFIER(:catalog_name || '.' || :schema_name || '.mask_video_id')(video_id) AS video_id,
    author,
    title,
    album
FROM IDENTIFIER(:catalog_name).IDENTIFIER(:schema_name).dim_music_metadata
WHERE IDENTIFIER(:catalog_name || '.' || :schema_name || '.author_row_filter')(author);


SELECT * from dbr_dev.music_analytics.cls_dim_music_metadata;
```
