```sql
CREATE OR REPLACE FUNCTION IDENTIFIER(:catalog_name).IDENTIFIER(:schema_name).author_row_filter(author_name STRING)
RETURNS BOOLEAN
RETURN
    is_account_group_member('admins')
    OR (is_account_group_member('linkin_park_fans') AND author_name = "Linkin Park")
    OR (is_account_group_member('metallica_fans') AND author_name = "Metallica");

-- We're bypassing the "materialized views don't support ALTER table" problem by creating an extra view
CREATE OR REPLACE VIEW IDENTIFIER(:catalog_name).IDENTIFIER(:schema_name).rls_dim_music_metadata AS
SELECT *
FROM IDENTIFIER(:catalog_name).IDENTIFIER(:schema_name).dim_music_metadata
WHERE IDENTIFIER(:catalog_name || '.' || :schema_name || '.author_row_filter')(author);

GRANT USE CATALOG ON CATALOG IDENTIFIER(:catalog_name) TO `linkin_park_fans`;
GRANT USE CATALOG ON CATALOG IDENTIFIER(:catalog_name) TO `metallica_fans`;

GRANT USE SCHEMA ON SCHEMA IDENTIFIER(:catalog_name).IDENTIFIER(:schema_name) TO `linkin_park_fans`;
GRANT USE SCHEMA ON SCHEMA IDENTIFIER(:catalog_name).IDENTIFIER(:schema_name) TO `metallica_fans`;

GRANT SELECT ON VIEW IDENTIFIER(:catalog_name).IDENTIFIER(:schema_name).rls_dim_music_metadata TO `linkin_park_fans`;
GRANT SELECT ON VIEW IDENTIFIER(:catalog_name).IDENTIFIER(:schema_name).rls_dim_music_metadata TO `metallica_fans`;
```
