```python
%sql
WITH dq_metrics AS (
    SELECT 
        (SELECT COUNT(*) FROM dbr_dev.music_analytics.silver_music_metadata) AS clean_count,
        (SELECT COUNT(*) FROM dbr_dev.music_analytics.silver_music_quarantine) AS quarantine_count
)
SELECT 
    clean_count,
    quarantine_count,
    (clean_count + quarantine_count) AS total_processed,
    ROUND((clean_count / (clean_count + quarantine_count)) * 100, 2) AS pass_rate_pct,
    CASE 
        WHEN quarantine_count = 0 THEN 'HEALTHY (100%)'
        WHEN (clean_count / (clean_count + quarantine_count)) >= 0.95 THEN 'WARNING (>95%)'
        ELSE 'CRITICAL (<95%)'
    END AS pipeline_status
FROM dq_metrics;
```
